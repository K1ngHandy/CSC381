```python
%pip install -q -U scikit-learn

import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error

df = pd.read_csv("AmesHousing_clean.csv")
features = ["Gr Liv Area", "Bedroom AbvGr", "Overall Qual", "House Age", "Garage Cars", "Total Bsmt SF",]
train = df[df["Yr Sold"] <= 2008]
valid = df[df["Yr Sold"] == 2009]

model = LinearRegression()
model.fit(train[features], np.log(train["SalePrice"]))

pred = np.exp(model.predict(valid[features]))
print(pd.Series(model.coef_, index=features).round(5))
print("Validation MAE:", round(mean_absolute_error(valid["SalePrice"], pred)))
print("Validation RMSE:", round(np.sqrt(mean_squared_error(valid["SalePrice"], pred))))
```
